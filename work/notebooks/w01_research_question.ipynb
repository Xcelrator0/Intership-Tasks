2. The question: decision, action, cost of a wrong call

Unit of Analysis: A single published web page (URL).

The Output: The predicted probability that updating the page will lead to a positive recovery in traffic/impressions.

The Decision & Action: The content manager decides which 10 decaying pages to assign to writers for a refresh this week.

Cost of a Wrong Call:

False Positive (recommending a page that fails to recover): Wasted 5–10 hours of editorial/writing budget.

False Negative (missing a high-potential page): Unrealized traffic growth and continued organic search decline.

Why Data/ML helps: Simple rules (such as "update articles older than 12 months") ignore complex signal combinations like position velocity, impressions, and CTR patterns.

3. Quick look at the data (2-3 real numbers)

```python
import pandas as pd
csv_path = "../../data/raw/content_refresh_anonymized.csv"
raw_url = "https://raw.githubusercontent.com/Xcelrator0/Intership-Tasks/main/data/raw/content_refresh_anonymized.csv"
try:
    df = pd.read_csv(csv_path)
    print(f"Loaded locally from: {csv_path}")
except Exception:
    df = pd.read_csv(raw_url)
    print(f"Loaded via GitHub URL: {raw_url}")
total_rows = len(df)
zero_imp_pct = (df['impressions_90d'] == 0).mean() * 100
median_position = df['avg_position'].median() if 'avg_position' in df.columns else None
print("\n--- DATA RESULTS ---")
print(f"1. Total pages in sample: {total_rows:,}")
print(f"2. Percentage of pages with 0 impressions over 90 days: {zero_imp_pct:.2f}%")
if median_position is not None:
    print(f"3. Median average position: {median_position:.2f}")
```
